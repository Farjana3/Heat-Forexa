# 03. Causal & Statistical Analysis

Correlation analysis, stationarity testing (ADF), and Granger causality analysis of environmental drivers.

```python
import sys
import os
sys.path.insert(0, os.path.abspath('..'))
sys.path.insert(0, os.path.abspath('.'))

import pandas as pd
import matplotlib.pyplot as plt
from src.analysis.causal import run_correlation_analysis, run_stationarity_tests, run_granger_causality_analysis
```

```python
# 1. Load Feature Dataset
df_feat = pd.read_csv('../data/processed/features.csv')
print(f'Features loaded: {df_feat.shape[0]:,} rows')
```

```python
# 2. Correlation Analysis
df_corr = run_correlation_analysis(df_feat, output_dir='output/result', fig_dir='output/figure')
print('Top correlations with target_t_plus_1:')
print(df_corr['target_t_plus_1'].sort_values(ascending=False))
```

```python
# 3. Stationarity Testing (Augmented Dickey-Fuller)
df_adf = run_stationarity_tests(df_feat, output_dir='output/result')
print('ADF Stationarity Test Results:')
print(df_adf.to_string(index=False))
```

```python
# 4. Granger Causality Testing
df_granger, selected_exog = run_granger_causality_analysis(df_feat, max_lag=5, output_dir='output/result', fig_dir='output/figure')
print('Granger Causality Results (p < 0.05):')
sig_granger = df_granger[df_granger['p_value'] < 0.05]
print(sig_granger.to_string(index=False) if len(sig_granger) > 0 else 'All candidate tests computed.')
print('\nSelected Exogenous Variables for SARIMAX:', selected_exog)
```
